| No. | Question to Practise | Bit Manipulation Concept |
|---:|---|---|
| 1 | Convert a decimal integer to binary. | Binary conversion |
| 2 | Check whether a number is odd or even. | `n & 1` |
| 3 | Check whether a number is a power of 2. | `n & (n - 1)` |
| 4 | Check whether a number is a power of 4. | Bit masking |
| 5 | Check whether the kth bit of a number is set. | `n & (1 << k)` |
| 6 | Set the kth bit of a number. | `n \| (1 << k)` |
| 7 | Unset (clear) the kth bit of a number. | `n & ~(1 << k)` |
| 8 | Toggle the kth bit of a number. | `n ^ (1 << k)` |
| 9 | Count the number of set bits (1s) in a number. | Bit counting |
| 10 | Remove the rightmost set bit of a number. | `n & (n - 1)` |

**Note:** Bit positions are zero-indexed, starting from the rightmost bit.

```python
def count_set_num(n):
    count = 0
    while n>0:
        count += (n&1)
        n = n>>1
    return count
```

# wirte a function wethaer a number is prime or not

```python
def check_prime(n):
    if n<=1:
        return False
    for i in range(2,n):
        if n%i==0:
            return False
    return True
```

```python
check_prime(11)
```
