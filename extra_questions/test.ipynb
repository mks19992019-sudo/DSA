```python
##
def merge(a1,a2):
    result = []
    i = 0
    j = 0
    while i < len(a1) and j < len(a2):
        if a1[i] < a2[j]:
            result.append(a1[i])
            i += 1
        else:
            result.append(a2[j])
            j += 1

    if i == len(a1):
        result.extend(a2[j:])
    else:
        result.extend(a1[i:])

    return result
    
    
```

```python
### merage sort

def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    n = len(arr)
    a1 = arr[:n//2]
    a2 = arr[n//2:]

    a1 = merge_sort(a1) 
    a2 = merge_sort(a2)
    return merge(a1, a2)

    
    
```
