# Rotate Array by One

## Approach
Single-step slicing

## Why I Used This Approach
I used slicing because rotating by one position can be expressed directly by taking the last element and placing the remaining elements after it.

## Time Complexity
`O(n)`

## Space Complexity
`O(n)`


```python
# ## Rotate Array by One

# given an array arr, rotate the array by one position in clockwise direction.

class Solution:
    def rotate(self, arr):
        arr[:] = arr[-1:] + arr[:-1]
        return arr
```
