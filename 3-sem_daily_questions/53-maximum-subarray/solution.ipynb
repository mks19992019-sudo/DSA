# 53. Maximum Subarray

## Approach
Kadane's Algorithm

## Why I Used This Approach
I used Kadane's Algorithm because it tracks the best subarray ending at the current index and gives the maximum sum in one pass.

## Time Complexity
`O(n)`

## Space Complexity
`O(1)`


```python
# ## 53. Maximum Subarray

# Given an integer array nums, find the subarray with the largest sum, and return its sum.

class Solution:
    def maxSubArray(self, nums: List[int]) -> int:
        max_sum = nums[0]
        current_sum=nums[0]
        for i in range(1,len(nums)):
            current_sum = max(nums[i],current_sum + nums[i])
            max_sum = max(max_sum,current_sum)
        return max_sum
```
