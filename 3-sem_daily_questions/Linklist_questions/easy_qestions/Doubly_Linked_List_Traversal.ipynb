### Doubly Linked List Traversal


```python
''' Structure of doubly linked list Node
  class Node:
      def __init__(self, x):
          self.data = x
          self.next = None
          self.prev = None
'''
class Solution:
    def displayList(self, head):
        # code here
        a= []
        b=[]
        temp = head
        
        while temp.next:
            val = temp.data
            a.append(val)
            temp = temp.next
        a.append(temp.data)
        
        while temp:
            b.append(temp.data)
            temp = temp.prev
        
        return [a,b]
        
```
