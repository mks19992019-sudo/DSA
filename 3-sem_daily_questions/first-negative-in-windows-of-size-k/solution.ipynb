# First Negative in Windows of Size K

## Approach
Sliding window with deque

## Why I Used This Approach
I used a deque to store useful negative indices only. This helps track the first negative element in each window without scanning the full window again.

## Time Complexity
`O(n)`

## Space Complexity
`O(k)`


```python
# ## First Negative in Windows of Size K

# brute force solution
class Solution:
    def firstNegInt(self, arr, k): 
         # code here 
        ans = []
        n = len(arr)
         
        for i in range(n-k+1):
            FOUND = False
            for j in range(i,i+k):
                if arr[j]<0:
                    ans.append(arr[j])
                    FOUND = True
                    break
            if not FOUND:
                ans.append(0)
        return ans

from collections import deque

class Solution:
    def firstNegInt(self, arr, k): 
         # code here 
        i = 0
        j = 0
        queue = deque()
        ans =[]
        
        while j<len(arr):
            
            if arr[j] < 0:
                queue.append(j)
            
            if (j-i+1) < k:
                j+=1
                
            elif (j-i+1) ==k:
                if queue:
                    ans.append(arr[queue[0]])
                else:
                    ans.append(0)
            
                if queue and queue[0] ==i:
                    queue.popleft()
                
                i +=1
                j+=1
                
        return ans

from collections import deque

class Solution:
    def firstNegInt(self, arr, k):
        queue = deque()
        ans = []

        for j in range(len(arr)):


            if arr[j] < 0:
                queue.append(j)


            if j >= k - 1:

                if queue:
                    ans.append(arr[queue[0]])
                else:
                    ans.append(0)


                if queue and queue[0] <= j - k:
                    queue.popleft()

        return ans
```
