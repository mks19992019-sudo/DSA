# 20. Valid Parentheses

## Approach
Stack-based matching

## Why I Used This Approach
I used a stack because parentheses must close in reverse order of opening, which matches stack behavior perfectly.

## Time Complexity
`O(n)`

## Space Complexity
`O(n)`


```python
# ## 20. Valid Parentheses

class Solution(object):
    def isValid(self, s):
        """
        :type s: str
        :rtype: bool
        """
        stack =[]
        check = {
            ')':'(',
            ']':'[',
            '}':'{'

        }

        for ch in s:
            if ch in '({[':
                stack.append(ch)
            else:
                if len(stack) is 0:
                    return False
                top = stack.pop()

                if top!=check[ch]:
                    return False
            
        return len(stack) == 0
```
