# Majority Element

## Approach
Boyer-Moore Voting Algorithm

## Why I Used This Approach
I used Boyer-Moore because it finds the majority element in one pass and without extra memory.

## Time Complexity
`O(n)`

## Space Complexity
`O(1)`


```python
# ## Majority Element

# it also done by hashmap

class Solution:
    def majorityElement(self, nums: List[int]) -> int:
        cadinate = None
        count = 0
        for num in nums:
            if count ==0:
                cadinate = num
            if num == cadinate:
                count +=1
            else:
                count -=1
        return cadinate
```
