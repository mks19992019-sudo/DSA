Given an array arr[]  and a range from l to r, the task is to count the number of triplets having a sum in the range [l, r].<br>

Examples :<br>

Input: arr = [8, 3, 5, 2], l = 7, r = 11<br>
Output: 1<br>
Explanation: There is only one triplet [2, 3, 5] having sum 10 in range [7, 11].<br>
Input: arr = [5, 1, 4, 3, 2], l = 2, r = 7<br>
Output: 2<br>
Explanation: There are two triplets having sum in range, [1,4,2] and [1,3,2].<br>


## 1 approch brute force 

```python
class Solution:
   
    def countTriplets(self, arr: list[int], l: int, r: int) -> int:
        # code here
        count = 0
        
        for i in range(len(arr)):
            for j in range(i+1,len(arr)):
                for k in range(j+1,len(arr)):
                    sum = arr[i] + arr[j] + arr[k]
                    if l <= sum <=r:
                        count+=1
        
        return count
            
# time complexcity O(o^3)
        
```
