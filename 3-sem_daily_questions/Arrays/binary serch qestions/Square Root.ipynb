```python
class Solution:
    def floorSqrt(self, n): 
        # code here
        
        i = 0
        j = n
        
        while i <=j:
            
            mid = (i+j) //2
            
            if mid**2 == n:
                return mid
            
            
            if mid**2 > n:
               
                j = mid -1
                
            else:
                result = mid
                i = mid + 1
                
            
        
        return result
                
                
        
            
                
        
        
            
        
                
            
            
        
```
