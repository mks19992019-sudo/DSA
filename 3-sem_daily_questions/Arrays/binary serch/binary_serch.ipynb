Given an array arr[], sorted in ascending order and an integer k. Return true if k is present in the array, otherwise, false.



```python
class Solution:
    def binarySearch(self, arr, k):
        # code here
        i = 0
        j = len(arr)-1
        
        while i <= j:
            mid = (i+j)//2
            if arr[mid] == k:
                return True
            if arr[mid] < k:
                i = mid +1
            else:
                j = mid -1
        
        return False
            
                
```
