# 287. Find the Duplicate Number

## Approach
Floyd's Cycle Detection

## Why I Used This Approach
I used Floyd's cycle method because the question asks for constant extra space. By treating values like pointers, we can find the duplicate without modifying the array.

## Time Complexity
`O(n)`

## Space Complexity
`O(1)`


```python
# ## 287. Find the Duplicate Number

# Given an array of integers nums containing n + 1 integers where each integer is in the range [1, n] inclusive.
#
# There is only one repeated number in nums, return this repeated number.
#
# You must solve the problem without modifying the array nums and using only constant extra space.

# # 1st

# without extra space

class Solution:
    def findDuplicate(self, nums: List[int]) -> int:
        slow = nums[0]
        fast = nums[0]

        while True:
            slow = nums[slow]
            fast = nums[nums[fast]]
            if slow == fast:
                break
        ptr = nums[0]
        while ptr != slow:
            slow = nums[slow]
            ptr = nums[ptr]
        return ptr

# ## 2nd

# with extra space

class Solution:
    def findDuplicate(self, nums: List[int]) -> int:
        seen = set()
        for i in nums:
            if i in seen:
                return i
            seen.add(i)

# # 3rd

# but the time complexity is O(n^2) and space complexity is O(1)

class Solution:
    def findDuplicate(self, nums: List[int]) -> int:
        nums.sort()
        for i in range(1,1+len(nums)):
            if nums[i] == nums[i-1]:
                return nums[i]
```
