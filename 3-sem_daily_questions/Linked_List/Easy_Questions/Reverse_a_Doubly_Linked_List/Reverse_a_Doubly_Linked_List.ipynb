## Reverse a Doubly Linked List

```python
""" Structure of Doubly Linked List Node
class Node:
    def __init__(self, val):
        self.data = val
        self.next = None
        self.prev = None
"""

class Solution:
    def reverse(self, head):
        # code here
        temp = head
        new_head = head
        
        while temp:
            temp.next , temp.prev = temp.prev , temp.next
            new_head = temp
            
            temp = temp.prev
            
        return new_head
            
        
            
         
        
        
```
