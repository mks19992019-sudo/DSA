```python
''' Structure of linked list Node
class Node:
    def __init__(self, x):
        self.data = x
        self.next = None
'''
class Solution:
    def pairwiseSwap(self, head):
        # code here
        first = head
        sec = head.next
        prev = None
        
        if (head ==None or head.next ==None):
            return head
        
        while(sec !=None):
            
            third = sec.next
            sec.next = first
            first.next = third
            
            if prev!=None:
                prev.next = sec
            else:
                head = sec
                
            prev = first
            
            first = third
            
            if third!=None:
            
                sec = third.next
            else:
                return head
            
        return head
        
            
            
            
            
```
