# 349. Intersection of Two Arrays

## Approach
Set lookup

## Why I Used This Approach
I used a set because membership checking becomes fast. This avoids nested loops and also helps keep only unique common elements.

## Time Complexity
`O(n + m)`

## Space Complexity
`O(n + m)`


```python
# ## 349. Intersection of Two Arrays

# Given two integer arrays nums1 and nums2, return an array of their intersection. Each element in the result must be unique and you may return the result in any order.

class Solution:
    def intersection(self, nums1: List[int], nums2: List[int]) -> List[int]:
        list1 = set(nums1)
        ans = set()

        for i in nums2:
            if i in list1:
                ans.add(i)

        return list(ans)
```
