## max 3 element of array

```python
arr = [2,5,-3,6,1,6,7]
max1 = float('-inf')
max2 = float('-inf')
max3 = float('-inf')


```

```python
for i in range(len(arr)):
    if max1<= arr[i]:
        max3=max2
        max2=max1
        max1 = arr[i]
    elif max2 <= arr[i]:
        max3=max2
        max2 = arr[i]
    elif max3 < arr[i]:
        max3 = arr[i]

print(max1)
print(max2)
print(max3)
```
