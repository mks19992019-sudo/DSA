## 232. Implement Queue using Stacks

```python
class MyQueue(object):

    def __init__(self):
        self.stack = []
        self.stack2 = []
        
    def push(self, x):
        """
        :type x: int
        :rtype: None
        """
        while self.stack:
            self.stack2.append(self.stack.pop())
        self.stack.append(x)
        while self.stack2:
            self.stack.append(self.stack2.pop())
       
    def pop(self):
        """
        :rtype: int
        """
        if self.empty():
            return 'queue is empty'
        return self.stack.pop()

    def peek(self):
        """
        :rtype: int
        """
        return self.stack[-1]

        

    def empty(self):
        """
        :rtype: bool
        """
        return len(self.stack)==0
        


# Your MyQueue object will be instantiated and called as such:
# obj = MyQueue()
# obj.push(x)
# param_2 = obj.pop()
# param_3 = obj.peek()
# param_4 = obj.empty()
```

## First Negative in Windows of Size K


```python
# brute force solution
class Solution:
    def firstNegInt(self, arr, k): 
         # code here 
        ans = []
        n = len(arr)
         
        for i in range(n-k+1):
            FOUND = False
            for j in range(i,i+k):
                if arr[j]<0:
                    ans.append(arr[j])
                    FOUND = True
                    break
            if not FOUND:
                ans.append(0)
        return ans
         
         
```
