```python
class Node:
    def __init__(self,data):
        self.val = data
        self.next = None

        
```

```python
class Queue:
    def __init__(self):
        self.front = None
        self.rear = None
    
    def enqueue(self,data):
        new_node = Node(data)
        if self.front is None:
            self.front = new_node
            self.rear = new_node
            return
        
        self.rear.next = new_node
        self.rear = new_node

    def dequeue(self):
        if self.front is None:
            return 'empty queue'
        
        removed = self.front.val
        if self.front.next is None:
            self.front = None
            self.rear = None
            return removed
        
        self.front = self.front.next
        return removed

    def peek(self):
        if self.front is None:
            return 'empty queue'
        return self.front.val
    
    def is_empty(self):
        return self.front is None
    
    def display(self):
        if self.front is None:
            return 'empty queue'
        
        temp = self.front
        while temp is not None:
            print(temp.val, end=' -> ')
            temp = temp.next
        
    def size(self):
        count = 0
        temp = self.front
        while temp is not None:
            count += 1
            temp = temp.next
        return count
        
    def clear(self):
        self.front = None
        self.rear = None
    
    def reverse(self):
        prev = None
        current = self.front
        while current is not None:
            next_node = current.next
            current.next = prev
            prev = current
            current = next_node
        self.front, self.rear = self.rear, self.front

    def check(self, traget ):
        if self.front is None:
            return 'empty queue'
        temp = self.front
        while temp is not None:
            if temp.val == traget:
                return 'found'
            temp = temp.next
        return 'not found'
    

    

```

```python
from collections import deque
```

```python
def bfs(root):
    result = []
    queue = deque()
    queue.append(root)
    while len(queue) != 0:
        e = queue.popleft()
        result.append(e.val)
        if e.left is not None:
            queue.append(e.left)
        if e.right is not None:
            queue.append(e.right)
    return result
```

```python
def pre_order(root):
    if root is None:
        return
    print(root.val, end=' ')
    pre_order(root.left)
    pre_order(root.right)
```

```python
pre_order(node_c)
```

```python
def in_order(root):
    if root is None:
        return
    in_order(root.left)
    print(root.val, end=' ')
    in_order(root.right)
```

```python
in_order(node_c)
```

```python
def post_order(root):
    if root is None:
        return
    post_order(root.left)
    post_order(root.right)
    print(root.val, end=' ')
```

```python
post_order(node_c)
```
