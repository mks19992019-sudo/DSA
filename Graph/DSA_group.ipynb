```python
adj_list ={
    8:[9,15],
    9:[28,8],
    28:[9,19],
    19:[28,102],
    102:[19,72],
    72:[102],
    15:[8,20],
    20:[15,50,58],
    50:[20],
    58:[20],
    72:[102]

}
```

```python
Visited = set()
```

```python
def DFS(node):
    if node in Visited:
        return
    Visited.add(node)
    print(node)

    for neberihods in  adj_list[node]:
        DFS(neberihods)

    


```

```python
from collections import deque

def bfs(scr):
    Visit = set()
    q = deque()
    q.append(scr)
    Visit.add(scr)

    ans_list = []
    

    while q:
        current_level =[]

        for i in range(len(q)):

            e = q.popleft()
            current_level.append(e)

            for neighbour in adj_list[e]:

                if neighbour not in Visit:
                    q.append(neighbour)
                    Visit.add(neighbour)
        ans_list.append(current_level)

    return ans_list
```

```python
bfs(8)
```
