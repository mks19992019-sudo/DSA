## 🌳 Binary Tree – Complete Notes

### Basic Definitions

- Node → a node is any element in a tree  

- Root node → the topmost node in a tree which does not have a parent  

- Parent node → a node that is an immediate predecessor of another node  

- Child node → a node that is an immediate successor of another node  

- Leaf node (external node) → a node that does not have any children  

- Ancestor node → any node above a given node (on the path from root)  

- Descendant node → any node below a given node  

- Sibling nodes → nodes that share the same parent  

- Level of a node → the number of edges from the root to that node  

- Internal node → a node which has at least one child  

- Neighbor node → the parent or child of a node  

- Subtree → a node and all its descendants form a tree  


---

### Key Relationships

- Ancestor and Descendant are opposite relations  

- If A is ancestor of B → B is descendant of A  

---

### Example Tree

```
        drink
       /     \
     hot     cold
    /  \     /   \
  TEA Coffee coco fenta
```

- Root → drink  
- Leaf nodes → TEA, Coffee, coco, fenta  
- Example ancestor → drink is ancestor of TEA  
- Example descendant → TEA is descendant of drink  
- Example siblings → hot and cold

## binary tree
Q. what is binary tree?
A. A binary tree is a type of data structure where:

 Each node can have at most two children

* One is called the left child
* The other is called the right child

Simple definition (easy to remember):

“A binary tree is a tree where every node has 0, 1, or 2 children.”

## 🌳 Types of Binary Trees

### 1. Binary Tree
**Definition:**  
A tree where each node can have 0, 1, or 2 children.

**Key Point:**  
No restriction on number of children (up to 2).

**Example:**
```
      A
     / \
    B   C
   /
  D
```

---

### 2. Full Binary Tree
**Definition:**  
A tree where every node has either 0 or 2 children.

**Key Point:**  
No node can have exactly 1 child.

**Example:**
```
      A
     / \
    B   C
   / \
  D   E
```

---

### 3. Complete Binary Tree
**Definition:**  
A tree where all levels are completely filled except possibly the last level, and the last level is filled from left to right.

**Key Point:**  
Nodes are filled level by level from left to right.

**Example:**
```
      A
     / \
    B   C
   / \
  D   E
```

---

### 4. Perfect Binary Tree
**Definition:**  
A tree where all internal nodes have 2 children and all leaf nodes are at the same level.

**Key Point:**  
- Every level is completely filled  
- Left to right filling happens automatically  

**Example:**
```
        A
      /   \
     B     C
    / \   / \
   D   E F   G
```

---

## 🔥 Quick Difference

Binary → 0, 1, or 2 children  
Full → only 0 or 2 children  
Complete → filled from left  
Perfect → completely filled  

---

## 🧠 Easy Trick

- Full → children rule  
- Complete → filling rule  
- Perfect → full + complete

## 🌳 Special Types of Binary Trees

### 1. Balanced Binary Tree
**Definition:**  
A binary tree in which the height difference between the left and right subtree of any node is at most 1.

**Key Point:**  
- Tree is “almost equal” on both sides  
- No side is too deep compared to the other  

**Example:**
```
        A
      /   \
     B     C
    / \     \
   D   E     F
```

👉 Height difference at every node ≤ 1 ✔️

---

### 2. Degenerate Binary Tree
**Definition:**  
A binary tree where each parent node has only one child.

**Key Point:**  
- Looks like a linked list  
- Tree becomes completely skewed  

**Example (Right Skewed):**
```
A
 \
  B
   \
    C
     \
      D
```

**Example (Left Skewed):**
```
      A
     /
    B
   /
  C
 /
D
```

---

## 🔥 Quick Difference

Balanced → height difference ≤ 1  
Degenerate → every node has only 1 child  

---

## 🧠 Easy Trick

- Balanced → “evenly spread”  
- Degenerate → “straight line (linked list)”

# trick

Step 1: Pick a node

Step 2: Go to LEFT child
        → Find longest downward path
        → Count edges

Step 3: Go to RIGHT child
        → Find longest downward path
        → Count edges

Step 4: Difference = |Left - Right|

Step 5:
    If diff ≤ 1 → Balanced ✔
    If diff > 1 → Not Balanced ❌
