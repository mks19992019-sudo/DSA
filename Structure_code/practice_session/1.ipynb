```python
class Node:
    def __init__(self,data):
        self.val = data
        self.next = None
        
```

```python
class link_list:
    def __init__(self):
        self.head = None
    
    def insert_head(self,data):
        temp = self.head 
        self.head = Node(data)

        self.head.next = temp


    def insert_at_end(self,data):
        new_node = Node(data)
        if self.head == None:
            self.head = new_node
            return
        
        temp = self.head

        while temp.next != None:
            temp = temp.next

        temp.next = new_node

    def traverse(self):
        if self.head == None:
            return 'empty list'
        
        temp = self.head

        while temp != None:
            print(temp.val , end = '->')

            temp = temp.next

    
    def reverse(self):
        cureent = self.head
        prev = None
        while cureent != None:
            next = cureent.next
            cureent.next = prev
            prev = cureent
            cureent = next
        self.head = prev



            




        
         

```

```python
class Node:
    def __init__ (self,data):
        self.val = data
        self.next = None


```

```python
class link_list:
    def __init__(self):
        self.head = None

    def add_song_start (self,data):
        new = Node(data)

        if self.head is None:
            self.head = new
            new.next = self.head
            return
        
        temp = self.head 
        while temp.next != self.head:
            temp = temp.next

        temp.next = new
        new.next = self.head
        self.head = new
        return 'song is added in the starting'
        
    def add_song (self,data):
        new = Node(data)
        if self.head is None:
            self.head = new
            new.next = self.head
            return
        temp = self.head
        while temp.next != self.head:
            temp = temp.next
        
        temp.next = new
        new.next = self.head
        return 'song is added'

    def show_songs (self):
        if self.head is None:
            return 'empty '
        temp = self.head
        while True:
            print(temp.val,end='- > ')
            temp = temp.next
            if temp is self.head:
                break

    def play_next(self,current_song):
        if self.head is None:
            return 'empty'
        temp = self.head
        while True:
            if temp.val == current_song:
                print('playing next song:',temp.next.val)
                break
            temp = temp.next
            if temp is self.head:
                print('song not found')
                break
    def delete_song(self,song_name):
        if self.head is None:
            return 'empty'
        temp = self.head
        while True:
            
            if temp.next.val == song_name:
                temp.next = temp.next.next
                break
            temp = temp.next
            if temp == self.head:
                print('song not found')
                break
            


```

```python
song = link_list()
```

```python
song.add_song('song1')
song.add_song('song2')
song.add_song('song3')  
```

```python
song.show_songs()
```

```python
song.play_next('song5')
```

```python
class dublley_link_list:
    def __init__(self):
        self.head = None
    def insert_head(self,data):
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            return
        self.head.prev = new_node
        new_node.next = self.head
        self.head = new_node

    def insert_end(self,data):
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            return
        
        temp = self.head

        while temp.next !=None:
            temp = temp.next
        
        new_node.prev = temp
        temp.next = new_node

    def display(self):
        if self.head is None:
            return 'empty list'
        temp = self.head

        while temp != None:
            print(temp.val, end=" < - > ")

            temp = temp.next
        

    def delete(self,target):
        if self.head is None:
            return 'empty list'
        if self.head.next == None:
            if self.head.val == target:
                self.head = None
                return
            else:
                return 'not found'
        if self.head.val == target:
            self.head = self.head.next
            self.head.prev = None
            return
        temp = self.head

        while temp.next != None:
            if temp.val == target:
                temp.prev.next = temp.next
                temp.next.prev = temp.prev
                return
            temp = temp.next
        if temp.val == target:
            temp.prev.next = None
        else:
            return 'not found'
        


    def insert(self, target ,data):
        new_node = Node(data)
        if self.head is None:
            return 'empty list'
        
        temp = self.head

        while temp.next != None:
            if temp.val == target:
                new_node.next = temp.next
                new_node.prev = temp
                temp.next.prev = new_node
                temp.next = new_node
                return
            temp = temp.next
        if temp.val == target:
            temp.next = new_node
            new_node.prev = temp
            return
        else:
            return 'not found'




              

        
        
            



```
