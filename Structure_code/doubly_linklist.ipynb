```python
class Node:
    def __init__(self, data):
        self.val = data
        self.next = None
        self.prev = None


class DoublyLinkedList:
    def __init__(self):
        self.head = None

    #  Insert at beginning
    def insert_head(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return
        
        new_node.next = self.head
        self.head.prev = new_node
        self.head = new_node

    #  Insert at end
    def insert_end(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return
        
        temp = self.head
        while temp.next is not None:
            temp = temp.next
        
        temp.next = new_node
        new_node.prev = temp

    # Insert after a value
    def insert_after(self, target, data):
        if self.head is None:
            return 'empty list'

        temp = self.head

        while temp is not None:
            if temp.val == target:

                new_node = Node(data)
                nxt = temp.next

                temp.next = new_node
                new_node.prev = temp
                new_node.next = nxt

                if nxt is not None:
                    nxt.prev = new_node

                return 'inserted'

            temp = temp.next

        return 'not found'


    def delete(self, target):
        if self.head is None:
            return 'empty list'

        # only one node
        if self.head.next is None:
            if self.head.val == target:
                self.head = None
                return 'deleted'
            else:
                return 'not found'

        # delete head
        if self.head.val == target:
            self.head = self.head.next
            self.head.prev = None
            return 'deleted'

        temp = self.head

        while temp.next is not None:
            if temp.val == target:
                temp.prev.next = temp.next
                temp.next.prev = temp.prev
                return 'deleted'
            temp = temp.next

        # last node
        if temp.val == target:
            temp.prev.next = None
            return 'deleted'

        return 'not found'

    #  Display forward
    def display(self):
        if self.head is None:
            print('empty list')
            return
        
        temp = self.head
        while temp is not None:
            print(temp.val, end=" <-> ")
            temp = temp.next
        print()

    #  Display reverse
    def display_reverse(self):
        if self.head is None:
            print('empty list')
            return
        
        temp = self.head
        while temp.next is not None:
            temp = temp.next
        
        while temp is not None:
            print(temp.val, end=" <-> ")
            temp = temp.prev
        print()

    #  String representation
    def __str__(self):
        if self.head is None:
            return 'DoublyLL is empty'
        
        result = ''
        temp = self.head
        while temp is not None:
            result += str(temp.val) + '<->'
            temp = temp.next
        
        return result[:-3]
```

```python
a = DoublyLinkedList()
```

```python
a.insert_after(3, 10)
```

```python
a.insert_end(3)
a.insert_end(2)
a.insert_end(6)
a.insert_end(7)
a.insert_end(5)
```

```python
a.display()
```

```python
a.display_reverse()
```
