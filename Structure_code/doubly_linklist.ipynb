```python
class Node:
    def __init__(self,val):
        self.val = val
        self.next = None
        self.prev = None
```

```python
class DoublyLL:
    def __init__(self):
        self.head = None

    def insert_last(self,value):
        New_node = Node(value)

        if self.head ==None:
            self.head = New_node
            return
        
        temp = self.head
        while temp.next != None:
            temp = temp.next

        temp.next = New_node
        New_node.prev = temp

    def __str__(self):
        if self.head ==None:
            return 'DoublyLL is empty'
        
        result = ''
        temp = self.head
        while temp != None:
            result = result + str(temp.val) + '<->'
            temp = temp.next
        
        return result[:-3]
    
    def insert_head(self,value):
        new_node = Node(value)
        temp = self.head
        if self.head == None:
            self.head = new_node
            return

        new_node.next = temp
        temp.prev = new_node
        self.head = new_node
    def insert(self,target,val):
        if self.head == None:
            return 'empty list'
        
        new_node = Node(val)
        temp = self.head
        while temp != None:
            if temp.val == target:
                break
            temp = temp.next
        if temp.val == target:
            next = temp.next
            temp.next = new_node
            new_node.prev = temp 
            new_node.next = next
        
        else:
            return 'elemnt not found'
    
    def remove(self,target):
        if self.head == None:
            return 'empty lits'
        if self.head.val == target:
            self.head = self.head.next
            self.head.prev = None
            return

        temp = self.head
        
        while temp.next != None:
            if temp.val== target:
                temp.prev.next = temp.next
                temp.next.prev = temp.prev
                return
            else:
                temp = temp.next
            
        if temp.val == target:
            temp.prev.next = None

        else:
            return 'element not found'

                

        



        

        
```
