```python
class Node:
    def __init__(self, val):
        self.data = val
        self.next =None
        self.prev = None
        
```

```python
class circular:
    def __init__(self):
        self.head = None

    def insert_at_beg(self, value):
        new_node = Node(value)

        if self.head == None:
            self.head = new_node
            new_node.next = self.head
            return
        
        temp = self.head
        while temp.next != self.head:
            temp = temp.next

        temp.next = new_node
        new_node.next = self.head
        self.head = new_node

    def insert_at_end(self, value):
        new_node = Node(value)

        if self.head == None:
            self.head = new_node
            new_node.next = self.head
            return
        
        temp = self.head
        while temp.next != self.head:
            temp = temp.next

        temp.next = new_node
        new_node.next = self.head

    def delete_from_begining(self):
        if self.head == None:
            return 'empty'
        
        if self.head.next == self.head:
            self.head = None
            return
        
        temp = self.head
        while temp.next != self.head:
            temp = temp.next

        temp.next = self.head.next
        self.head = self.head.next

    def delete_from_end(self):
        if self.head == None:
            return 'empty'
        
        if self.head.next == self.head:
            self.head = None
            return
        
        temp = self.head
        while temp.next.next != self.head:
            temp = temp.next

        temp.next = self.head

    def search(self, value):
        if self.head == None:
            return False
    
        temp = self.head
        while True:
            if temp.data == value:
                return True
        
            temp = temp.next
            if temp == self.head:
                return False

    def length(self):
        if self.head == None:
            return 0
    
        count = 0
        temp = self.head

        while True:
            count += 1
            temp = temp.next
            if temp == self.head:
                break

        return count

    def traverse(self):
        if self.head == None:
            print('empty')
            return
    
        temp = self.head
        while True:
            print(temp.data, end=" -> ")
            temp = temp.next
            if temp == self.head:
                break

        print("(back to head)")
```
