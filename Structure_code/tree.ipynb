# different traversal in a binary tree

1. DFS (Depth First Search) Traversal:
   - Inorder Traversal: Left, Root, Right
   - Preorder Traversal: Root, Left, Right
   - Postorder Traversal: Left, Right, Root

2. BFS (Breadth First Search) Traversal:
    - Level Order Traversal: Traverse the tree level by level from left to right.

### hack for order of traversal

- inorder mean root in middle
- preorder mean root in first
- postorder mean root in last

```python
class Node:
    def __init__(self,data):
        self.val = data
        self.left = None
        self.right = None
    
    
```

```python
# random tree making 
node1 = Node(1)
node2 = Node(2)
node3 = Node(3)
node4 = Node(4)
node5 = Node(5)
node6 = Node(6)
node7 = Node(7)


node1.left = node2
node1.right = node3
node2.left = node4
node2.right = node5
node3.left = node6
node3.right = node7

```

```python
def preorder_traversal(root):
    if root is None:
        return
    # print the value of the current node
    print(root.val , end=' ')

    # Traverse the left subtree
    preorder_traversal(root.left)

    # Traverse the right subtree
    preorder_traversal(root.right)



    
```

```python
preorder_traversal(node1)
```

```python
def inorder_traversal(root):
    if root is None:
        return
    

    inorder_traversal(root.left)
    print(root.val , end=' ')
    inorder_traversal(root.right)
    
    
```

```python
inorder_traversal(node1)
```

```python
def postorder_traversal(root):
    if root is None:
        return
    
    postorder_traversal(root.left)
    postorder_traversal(root.right)
    print(root.val , end=' ')
```

```python
postorder_traversal(node1)
```

## BFS (Level Order Traversal)

```python
# tree

class tree:
    def __init__(self,data):
        self.val = data
        self.left = None
        self.right = None
        
```

```python
# tree nodes
node1 = tree('A')
node2 = tree('B')
node3 = tree('C')
node4 = tree('D')
node5 = tree('E')
node6 = tree('F')
node7 = tree('G')

# tree structure
node1.left = node2
node1.right = node3
node2.left = node4
node2.right = node5
node3.left = node6
node3.right = node7


```

```python
from collections import deque
```

```python
def level_order(node1):
    result = []
    queue = deque()
    queue.append(node1)
    while len(queue) != 0:
        e = queue.popleft()
        result.append(e.val)
        if e.left is not None:
            queue.append(e.left)
        if e.right is not None:
            queue.append(e.right)
    return result
```

```python
level_order(node1)
```
