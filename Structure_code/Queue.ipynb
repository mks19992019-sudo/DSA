# queue using linked list

insert => enqueue (here we are inserting the element at the end of the queue)
delete => dequeue (here we are deleting the first element of the queue)

```python
class Node:
    def __init__(self,value):
        self.data = value
        self.next = None

    
```

```python
class Queue:
    def __init__(self):
        self.front = None
        self.rear = None

    def enqueue(self,value):
        new_node = Node(value)

        if self.rear == None:
            self.front = new_node
            self.rear = self.front
        
        else:
            # now we add the node at the end ok queue 
            self.rear.next = new_node
            # in this we change the rear to the new node
            self.rear = new_node
    def dequeue(self):
        if self.front == None:
            return 'queue is empty'
        
        if self.front.next ==None:
            self.front = None
            self.rear = None
            return

        self.front = self.front.next

        
    def front_(self):
        if self.front == None:
            return 'empty'
        return self.front.data
    
    def rear_(self):
        if self.front == None:
            return 'empty'
        return self.rear.data
    
    def is_empty(self):
        return  self.front == None
    
    def count(self):
        if self.front == None:
            return 'empty 0 '
        count = 0
        temp = self.front

        while temp != None:
            count = count + 1
            temp = temp.next
            
        return count

    def __str__(self):
        if self.front==None:
            return 'empty'
        
        else:
            result =''
            temp = self.front

            while temp !=None:
                result = result + str(temp.data) + '->'

                temp = temp.next
            
            return result[:-2]
```

```python
queue.enqueue(5)
queue.enqueue(6)
queue.enqueue(7)
```

# priorty queue

```python
import heapq
```

```python
# min priority queue
class minPriorityQueue:
    def __init__(self):
        self.queue = []
    
    def enqueue(self,value):
        heapq.heappush(self.queue,value)

    def dequeue(self):
        if len(self.queue) == 0:
            return 'queue is empty'
        return heapq.heappop(self.queue)
    def peek(self):
        if len(self.queue) == 0:
            return 'queue is empty'
        return self.queue[0]
    def is_empty(self):
        return len(self.queue) == 0
    def size(self):
        return len(self.queue)
    def __str__(self):
        return str(self.queue)
    
```

```python
min_pq = minPriorityQueue()
```

```python
min_pq.enqueue(5)
min_pq.enqueue(3)
min_pq.enqueue(8)
```

```python
print(min_pq)
```

```python
min_pq.dequeue()
```

# max priorty queue

```python
class maxPriorityQueue:
    def __init__(self):
        self.queue = []
    
    def enqueue(self,value):
        heapq.heappush(self.queue,-value)
    
    def dequeue(self):
        if len(self.queue) == 0:
            return 'queue is empty'
        return -heapq.heappop(self.queue)
    
    def __str__(self):
        return str(self.queue)
    
```

```python
max = maxPriorityQueue()
```

```python
max.enqueue(5)      
max.enqueue(3)
max.enqueue(8)
```

```python
print(max)
```

```python
max.dequeue()
```

# random priorty queue

```python
class PriorityQueue:
    def __init__(self):
        self.e = []
    
    def enqueue(self,value,priority):
        heapq.heappush(self.e,(priority,value))
    def dequeue(self):
        if len(self.e) == 0:
            return 'empty queue'
        return heapq.heappop(self.e)[1]
    
    def __str__(self):
        return str(self.e)
```

```python
random = PriorityQueue()
```

```python
random.enqueue('A',2)
random.enqueue('B',1)
random.enqueue('C',10)
```

```python
print(random)
```

```python
random.dequeue()
```
