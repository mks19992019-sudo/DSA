# queue using linked list

insert => enqueue (here we are inserting the element at the end of the queue)
delete => dequeue (here we are deleting the first element of the queue)

```python
class Node:
    def __init__(self,value):
        self.data = value
        self.next = None

    
```

```python
class Queue:
    def __init__(self):
        self.front = None
        self.rear = None

    def enqueue(self,value):
        new_node = Node(value)

        if self.rear == None:
            self.front = new_node
            self.rear = self.front
        
        else:
            # now we add the node at the end ok queue 
            self.rear.next = new_node
            # in this we change the rear to the new node
            self.rear = new_node
    def dequeue(self):
        if self.front == None:
            return 'queue is empty'
        
        if self.front.next ==None:
            self.front = None
            self.rear = None
            return

        self.front = self.front.next

        
    def front_(self):
        if self.front == None:
            return 'empty'
        return self.front.data
    
    def rear_(self):
        if self.front == None:
            return 'empty'
        return self.rear.data
    
    def is_empty(self):
        return  self.front == None
    
    def count(self):
        if self.front == None:
            return 'empty 0 '
        count = 0
        temp = self.front

        while temp != None:
            count = count + 1
            temp = temp.next
            
        return count

    def __str__(self):
        if self.front==None:
            return 'empty'
        
        else:
            result =''
            temp = self.front

            while temp !=None:
                result = result + str(temp.data) + '->'

                temp = temp.next
            
            return result[:-2]
```
