## build stack using list


```python
class Stack:
    def __init__(self):
        self.s=[]

    def length(self):
        return len(self.s)
    
    def push(self,value):
        self.s.append(value)
    
    
    def peek(self):
        if len(self.s) == 0:
            raise Exception('stack is empty')
        
        return self.s[-1]
    
    def pop(self):
        if len(self.s)==0:
            raise Exception('empty list')
        return self.s.pop()
        
```

```python
a = Stack()
```

```python
a.push(4)
a.push(5)
```

```python
a.length()
```

```python
a.pop()
```

```python
a.peek()
```

```python
class Node:
    def __init__(self,value):
        self.val = value
        self.next = None

    
```

```python
class stack:
    def __init__(self):
        self.top = None

    def isEmpty(self):
        return self.top == None
    
    def push(self,val):
        new_node = Node(val)

        new_node.next = self.top
        self.top = new_node

    def peek(self):
        if (self.isEmpty()):
            return 'stack is empty'
        return self.top.val
    
    def trace (self):
        if (self.isEmpty()):
            return 'stack is empty'
        temp = self.top

        while temp != None:
            print(temp.val)
            temp = temp.next
        
    def pop(self):
        if (self.isEmpty()):
            return 'stack is empty'
        data = self.top.val
        self.top = self.top.next
        return(data)
        

            
        

```

#### Q1.reverse the string using stack

```python
def reverse_string(text):
    b = stack()
    
    for i in text:
        b.push(i)

    res = ''

    while (not b.isEmpty()):
        res = res + str(b.pop())

    return res

```

```python
reverse_string('mohit')
```
