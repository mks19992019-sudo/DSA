Q1. Given five positive integers, find the minimum and maximum values that can be calculated by summing exactly four of the five integers. Then print the respective minimum and maximum values as a single line of two space-separated long integers.

```python
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'miniMaxSum' function below.
#
# The function accepts INTEGER_ARRAY arr as parameter.
#

def miniMaxSum(arr):
    # Write your code here
    a = arr
    sum = 0
    min_value = a[0]
    max_value = a[0]
    
    for i in range(len(a)):
        sum = sum + a[i]
        if min_value > a[i]:
            min_value = a[i]
        if max_value < a[i]:
            max_value = a[i]
    max_sum = sum - min_value
    min_sum = sum - max_value
    print(min_sum,end=" ")
    print(max_sum)
    
   

if __name__ == '__main__':
    

    arr = list(map(int, input().rstrip().split()))

    miniMaxSum(arr)


"""perf: optimize miniMaxSum using single-pass traversal

- Calculate total sum, minimum, and maximum values in one loop

- Avoid separate min() and max() scans

- Compute minSum and maxSum in O(n) time with O(1) space"""
```

leetcode 189. Rotate Array

Q. Given an integer array nums, rotate the array to the right by k steps, where k is non-negative.

```python
class Solution:
    def rotate(self, nums: List[int], k: int) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        n = len(nums)
        k %= n
        nums[:] = nums[-k:] + nums[:-k]
        
```

leetcode 2439. Minimize Maximum of Array

Q-> You are given a 0-indexed array nums comprising of n non-negative integers.

In one operation, you must:

Choose an integer i such that 1 <= i < n and nums[i] > 0.
Decrease nums[i] by 1.
Increase nums[i - 1] by 1.
Return the minimum possible value of the maximum integer of nums after performing any number of operations


```python
class Solution:
    def minimizeArrayValue(self, nums: List[int]) -> int:
        prefix_sum = 0
        ans = 0

        for i , num in enumerate(nums):
            prefix_sum += num
            avg = (prefix_sum + i) // (i+1)

            ans = max(ans, avg)
        return ans

        
```

leetcode 215. Kth Largest Element in an Array

q- Given an integer array nums and an integer k, return the kth largest element in the array.
Note that it is the kth largest element in the sorted order, not the kth distinct element.
Can you solve it without sorting?

```python
import heapq
class Solution:
    def findKthLargest(self, nums: List[int], k: int) -> int:
        heap = []
        for i in nums:
            heapq.heappush(heap,i)

            if len(heap) > k:
                heapq.heappop(heap)
        return heap[0]
        
```

leetcode 53. Maximum Subarray

Given an integer array nums, find the subarray with the largest sum, and return its sum.

```python
# this algo is very slow but good for starting 
class Solution:
    def maxSubArray(self, nums: List[int]) -> int:
        max_sum = float('-inf')

        for i in range(len(nums)):
            current_sum = 0
            for j in range (i,len(nums)):
                current_sum += nums[j]
                max_sum = max(current_sum , max_sum)
        return max_sum
        
```

```python
# more optmize code
class Solution:
    def maxSubArray(self, nums: List[int]) -> int:
        current_sum = nums[0]

        max_sum = nums[0]

        for i in range(1, len(nums)):

            current_sum = max(nums[i], current_sum + nums[i])

            max_sum = max(max_sum, current_sum)

        return max_sum
```

287. Find the Duplicate Number (leetcode)

Given an array of integers nums containing n + 1 integers where each integer is in the range [1, n] inclusive.

There is only one repeated number in nums, return this repeated number.

You must solve the problem without modifying the array nums and using only constant extra space.

 





```python
class Solution:
    def findDuplicate(self, nums: List[int]) -> int:
        slow = nums[0]
        fast = nums[0]

        while True:
            slow = nums[slow]
            fast= nums[nums[fast]]

            if slow == fast:
                break

        ptr = nums[0]
        while ptr != slow:
            ptr = nums[ptr]
            slow = nums[slow]
        return ptr


        
```

88. Merge Sorted Array

```python
# first time 
class Solution:
    def merge(self, nums1: List[int], m: int, nums2: List[int], n: int) -> None:
        """
        Do not return anything, modify nums1 in-place instead.
        """
        nums1[:] = nums1[:m] + nums2
        nums1 = nums1.sort()
```

```python
# second approch
class Solution:
    def merge(self, nums1: List[int], m: int, nums2: List[int], n: int) -> None:
        """
        Do not return anything, modify nums1 in-place instead.
        """
        i = m - 1
        j = n - 1
        k = m + n -1
        while i >= 0 and j >= 0 :
            if nums1[i] > nums2[j]:
                nums1[k] = nums1[i]
                i-=1
            else:
                nums1[k] = nums2[j]
                j-=1
            k-=1
          
        while j>=0:
            nums1[k] = nums2[j]
            j-=1
            k-=1
```
