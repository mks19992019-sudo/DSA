# impliment stack using queue 

-> queue (FIFO) and stack (LIFO) are two different data structures. However, we can use a queue to implement a stack by following these steps:

its actually behave like stack (LIFO) but we use queue to impliment it

```python
from collections import deque
```

```python
class stackUsingDeque:
    def __init__(self):
        self.queue = deque()

    def push(self,val):
        self.queue.append(val)

        for i in range (len(self.queue)-1):
            self.queue.append(self.queue.popleft())
        
    def pop(self):
        if self.is_empty():
            return 'empty stack'

        return self.queue.popleft()

    def peek(self):
        if self.is_empty():
            return 'empty stack'
        return self.queue[0]

    def is_empty(self):
        return len(self.queue) == 0

    def size(self):
        return len(self.queue)
```

```python
b = stackUsingDeque()
```
