```python
class Node:
    def __init__(self,val):
        self.val = val
        self.left = None
        self.right = None
        
```

```python
node1 = Node(8)
node2 = Node(10)
node3 = Node(11)
node4 = Node(12)
node5 = Node(2)
node6 = Node(4)
node7 = Node(3)
```

```python
node1.right = node2
node1.left = node3
node3.left  = node4
node3.right = node5
node2.left = node6
node2.right = node7
```

```python
def print_root(root):
    if root ==None:
        return 
    print(root.val)
    print_root(root.left)
    print_root(root.right)
```

```python
print_root(node1)
```

```python
def print_mid(root):
    if root ==None:
        return 
    
    print_root(root.left)
    print(root.val)
    print_root(root.right)
```

```python
print_mid(node1)
```

```python
def print_end(root):
    if root ==None:
        return 
    
    print_root(root.left)
    
    print_root(root.right)
    print(root.val)
```

```python
print_end(node1)
```

```python
def serch_tree(root,k):
    if root == None:
        return
    if root.val == k:
        print( 'elemnt found')

    serch_tree(root.left,k)
    serch_tree(root.right,k)
    return False

    
```

```python
serch_tree(node1,111)
```

```python
def node_sum2(root):
    if root == None:
        return 0

    sum = root.val

    sum += node_sum(root.left)
    sum += node_sum(root.right)

    return sum
```

```python
def node_sum(root):
    if root is None:
        return 0

    return root.val + node_sum(root.left) + node_sum(root.right)
```

```python
node_sum2(node1)
```
