```python
class Linklists:
    def __init__(self):
        self.head = None

    def head_insert(self,value):
        new_node = Node(value)
        new_node.next = self.head
        # now head is new_node
        self.head = new_node

    def __str__(self):
        if self.head == None:
            return 'empty list'
        temp = self.head
        result = ''
        while temp != None:
            result = result + str(temp.val) + '->'
            temp = temp.next
        return result[:-2]
    
    def apend(self,value):
        new_node=Node(value)
        if self.head == None:
            self.head = new_node
            return


        temp = self.head

        while temp.next != None:
            temp = temp.next

        temp.next = new_node
    # max value in linklist
    def max(self):
        temp = self.head
        max = self.head.val
        while temp != None:
            if temp.val > max:
                max = temp.val
            temp = temp.next
        
        return f'max value is {max}'

    # reverse of link_list

    def revers(self):
        current = self.head
        previous = None

        while current!=None:
            next = current.next
            current.next = previous

            previous = current

            current = next

        self.head = previous



```
