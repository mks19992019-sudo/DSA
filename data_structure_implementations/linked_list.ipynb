```python

class Linklist:
    def __init__(self):
        self.head =None    

    def head_insert(self,value):
        new_node = Node(value)
        new_node.next = self.head
        self.head = new_node

    def __str__(self):
        if self.head ==None:
            return 'empty linklist'
        temp = self.head
        result = ''
        while temp !=None:
            result = result + str(temp.val) + '->'
            temp = temp.next
        result = result + 'None'
        return result

    def last_add(self,val):
        new_node = Node(val)
        if self.head ==None:
            self.head =new_node
            return 

        temp = self.head
        while temp.next != None:
            temp = temp.next

        temp.next = new_node

    def insert_index(self, index, val):
        new_node = Node(val)
        if index == 0:
            new_node.next = self.head
            self.head = new_node
            return 

        temp = self.head
        count = 0

        while temp and count < index-1 :
            temp = temp.next
            count += 1
        if temp ==None:
            return 'index out of range'

        new_node.next = temp.next
        temp.next = new_node

    def insert_target(self,target,val):
        new_node = Node(val)
        temp = self.head

        while temp!=None:
            if temp.val == target:
                break
            temp = temp.next
        if temp!=None:
            next = temp.next
            temp.next = new_node
            new_node.next = next
        else:
            return 'element not found'

    def head_pop(self):

        if self.head ==None:
            return 'empty linklist'
        self.head = self.head.next

    def pop(self):
        if self.head ==None:
            return 'empty linklist'
        if self.head.next ==None:
            return self.head_pop()
        
        temp = self.head
        while temp.next.next!= None:
            temp = temp.next

        temp.next = None

    def clear(self):
        self.head =None    

    def delete(self, target):
        if self.head ==None:
            return 'empty linklist'
        if self.head.val == target:
            self.head = self.head.next
            return         
        temp = self.head
        
        while temp.next != None:
            if temp.next.val == target:
                temp.next = temp.next.next
                return
               
            temp = temp.next
        return 'target not found'

    def serch(self,target):
        if self.head ==None:
            return 'empty linklist'
        temp = self.head
        count = 0


        while temp!=None:
            
            if temp.val == target:
                return f"value is at index {count}"
            count +=1
            temp = temp.next
        return 'target not found'

    def max(self):
        if self.head ==None:
            return 'empty linklist'
        max = -float('inf')

        temp = self.head

        while temp != None:
            if temp.val > max:
                max = temp.val
            temp= temp.next
        return max

    def reverse(self):
        prev = None
        current = self.head

        while current != None:
            next = current.next
            current.next = prev
            prev = current
            current = next

        self.head = prev
        return




        

            
            


        

```
