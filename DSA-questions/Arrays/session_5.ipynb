70. Climbing Stairs

You are climbing a staircase. It takes n steps to reach the top.

Each time you can either climb 1 or 2 steps. In how many distinct ways can you climb to the top?

Problem: Climbing Stairs

Initial Thought Process:
- I started by observing that to reach the nth stair, the last move can only come from:
  - stair (n - 1) by taking 1 step
  - stair (n - 2) by taking 2 steps
- This leads to the recurrence relation:

  f(n) = f(n - 1) + f(n - 2)

- Base cases:
  - f(1) = 1
  - f(2) = 2

First Attempt:
- I implemented a recursive solution directly using the recurrence relation.
- Although the logic was correct, it resulted in repeated computation of the same subproblems and exceeded the time limit for larger inputs.

Observation:
- While tracing the recursion tree, I noticed that values such as f(3), f(4), etc., were being calculated multiple times.
- This revealed the presence of overlapping subproblems, making the problem a good candidate for Dynamic Programming.

Optimization:
- I introduced memoization using a dictionary.
- Before computing a state, I check whether it has already been solved.
- If the result exists in the memo dictionary, it is returned immediately.
- Otherwise, the result is computed, stored, and reused whenever needed.

Key Mistake:
- Initially, I called self.climbStairs(n - 1) and self.climbStairs(n - 2) inside the memoized helper function.
- This bypassed the memoization logic.
- The correct approach is to call the helper function itself (dfs/rc) so that previously computed states can be reused.

Time Complexity: O(n)
Space Complexity: O(n)

```python
class Solution:
    def climbStairs(self, n: int) -> int:
        memo = {}
        def rc(n):
            if n == 1:
                return 1
            if n == 2:
                return 2
            if n in memo:
                return memo[n]
            
            memo[n] =  rc(n-1) + rc(n-2)

            return memo[n]

        return rc(n)


        
```

171. Excel Sheet Column Number

Given a string columnTitle that represents the column title as appears in an Excel sheet, return its corresponding column number.

For example:

A -> 1
B -> 2
C -> 3
...
Z -> 26
AA -> 27
AB -> 28 
...
 

Example 1:

Input: columnTitle = "A"
Output: 1

solution: approach -> Convert the column title to a number by treating it as a base-26 number system.

```python
class Solution:
    def titleToNumber(self, columnTitle: str) -> int:
        ans = 0

        for ch in columnTitle:

            value = ord(ch) - ord("A") + 1
            ans = ans*26 + value

        return ans
```

202. Happy Number

Write an algorithm to determine if a number n is happy.

A happy number is a number defined by the following process:

Starting with any positive integer, replace the number by the sum of the squares of its digits.
Repeat the process until the number equals 1 (where it will stay), or it loops endlessly in a cycle which does not include 1.
Those numbers for which this process ends in 1 are happy.
Return true if n is a happy number, and false if not.

 

Example 1:

Input: n = 19
Output: true
Explanation:
12 + 92 = 82
82 + 22 = 68
62 + 82 = 100
12 + 02 + 02 = 1

```python
class Solution:
    def isHappy(self, n: int) -> bool:
        number = n
        sum = 0 
        memo = set()

        while number > 0:
            sum = 0
            while number>0:
                digit = number%10
                sum = digit*digit + sum
                number = number//10
            if sum ==1:
                return True
            
            if sum in memo:
                return False  
            else:
                memo.add(sum)
            number = sum
            

        
```

231. Power of Two

Given an integer n, return true if it is a power of two. Otherwise, return false.

An integer n is a power of two, if there exists an integer x such that n == 2x.

 

Example 1:

Input: n = 1
Output: true
Explanation: 20 = 1
Example 2:

Input: n = 16
Output: true
Explanation: 24 = 16
Example 3:

Input: n = 3
Output: false

```python

class Solution:
    def isPowerOfTwo(self, n: int) -> bool:
        if n<=0:
            return False
        while n>1:
            if n%2 !=0:
                return False
            n = n//2

        return True
            
```

258. Add Digits

Given an integer num, repeatedly add all its digits until the result has only one digit, and return it.

 

Example 1:

Input: num = 38
Output: 2
Explanation: The process is
38 --> 3 + 8 --> 11
11 --> 1 + 1 --> 2 
Since 2 has only one digit, return it.

```python
class Solution:
    def addDigits(self, num: int) -> int:
        if 0<= num < 10:
            return num

        while num >= 10:
            sum = 0

            while num > 0:
                
                digit = num%10
                sum = sum + digit

                num = num //10

            num = sum

        return num

                

            
```

263. Ugly Number

An ugly number is a positive integer which does not have a prime factor other than 2, 3, and 5.

Given an integer n, return true if n is an ugly number.

 

Example 1:

Input: n = 6
Output: true
Explanation: 6 = 2 × 3
Example 2:

Input: n = 1
Output: true
Explanation: 1 has no prime factors.
Example 3:

Input: n = 14
Output: false
Explanation: 14 is not ugly since it includes the prime factor 7.


```python
class Solution:
    def isUgly(self, n: int) -> bool:
        if n <= 0:
            return False
        while n%2==0:
            n = n//2
        while n % 3 == 0:
            n = n//3
        while n% 5 ==0:
            n = n // 5
        return n == 1

        
```

# 34. Find First and Last Position of Element in Sorted Array

## Problem Statement

Given a sorted array nums and a target value, return the first and last position of the target in the array. If the target does not exist, return [-1, -1].

Constraint: The solution must run in O(log n) time complexity.

---

## My Initial Thought Process

My first idea was to perform a standard Binary Search to locate any occurrence of the target.

Once the target is found:

1. Perform another Binary Search on the left half to find the first occurrence.
2. Perform another Binary Search on the right half to find the last occurrence.
3. Return both indices.

### Visualization

text nums = [5,7,7,8,8,10] target = 8            ↓ Binary Search finds 8 at index 4  Left Side Search: [5,7,7,8]        ↑ first occurrence = 3  Right Side Search: [8,10]  ↑ last occurrence = 4  Answer = [3,4] 

### Complexity

- Initial Binary Search: O(log n)
- Left Boundary Search: O(log n)
- Right Boundary Search: O(log n)

Overall:

text O(log n) 

because constants are ignored in Big-O notation.

---

## Optimized Approach

After further analysis, I realized that the initial Binary Search can be removed completely.

Instead:

### Binary Search #1 → First Occurrence

Whenever the target is found:

text Store answer Move to the left half 

This continues until the leftmost occurrence is found.

### Binary Search #2 → Last Occurrence

Whenever the target is found:

text Store answer Move to the right half 

This continues until the rightmost occurrence is found.

### Visualization

text nums = [5,7,7,8,8,10]  First Occurrence Search  Target found at index 4 Store 4 Search Left  Target found at index 3 Store 3 Search Left  No more targets First = 3 

text nums = [5,7,7,8,8,10]  Last Occurrence Search  Target found at index 4 Store 4 Search Right  No more targets Last = 4 

### Complexity

- First Occurrence Search: O(log n)
- Last Occurrence Search: O(log n)

Overall:

text O(log n) 

Space Complexity:

text O(1) 

---

## Key Learning

This problem reinforced an important Binary Search pattern:

text Target found ≠ Stop searching 

For boundary-related problems:

- First Occurrence → Continue searching left.
- Last Occurrence → Continue searching right.

This pattern is also useful for problems involving:

- Lower Bound
- Upper Bound
- Range Queries
- Boundary Search Problems

```python
# first approch

class Solution:
    def searchRange(self, nums: List[int], target: int) -> List[int]:
        left = 0
        right = len(nums) -1 
        mid = 0

        while  left <= right:
            mid = (left+right)//2

            if nums[mid]== target:
                first_idx = mid
                last_idx = mid
                                 
                # for left 
                right1 = mid -1
                left1 = left
                while left1 <=right1:
                    mid1 = (right1+left1)//2
                    if nums[mid1] == target:
                        first_idx = mid1
                        right1 = mid1 -1
                        
                    elif nums[mid1] < target:
                        left1 = mid1+1
                    else :
                        right1 = mid1 -1
                
                # for right 
                right2 = right 
                left2 = mid +1
                while left2 <= right2 :
                    mid2 = (right2+left2)//2
                    if nums[mid2] == target:
                        last_idx = mid2
                        left2 = mid2+1
                    elif nums[mid2] > target:
                        right2 = mid2 -1
                    else:
                        left2= mid2-1
                return [first_idx,last_idx]

            elif nums[mid] < target:
                left = mid +1

            else:
                right = mid - 1

        return [-1,-1]

        
```

```python
# second approch

class Solution:
    def searchRange(self, nums: List[int], target: int) -> List[int]:
        left = 0
        right = len(nums)-1
        first =-1
        # for left 
        while left<=right:
            mid = (left+right)//2
            if nums[mid] == target:
                first = mid
                right = mid -1
            elif nums[mid] <= target:
                left = mid+1
            else :
                right = mid -1
        
        last = -1
        left = 0
        right = len(nums)-1
        # for right 
        while left<=right:
            mid = (left+right)//2
            if nums[mid] == target:
                last = mid
                left = mid +1
            elif nums[mid] >= target:
                right = mid-1
            else :
                left = mid +1
        
        return [first,last]




        
```

153. Find Minimum in Rotated Sorted Array

Suppose an array of length n sorted in ascending order is rotated between 1 and n times. For example, the array nums = [0,1,2,4,5,6,7] might become:

[4,5,6,7,0,1,2] if it was rotated 4 times.
[0,1,2,4,5,6,7] if it was rotated 7 times.
Notice that rotating an array [a[0], a[1], a[2], ..., a[n-1]] 1 time results in the array [a[n-1], a[0], a[1], a[2], ..., a[n-2]].

Given the sorted rotated array nums of unique elements, return the minimum element of this array.

You must write an algorithm that runs in O(log n) time.

 

Example 1:

Input: nums = [3,4,5,1,2]
Output: 1
Explanation: The original array was [1,2,3,4,5] rotated 3 times.

```python
class Solution:
    def findMin(self, nums: List[int]) -> int:
        left = 0
        right = len(nums)-1

        while left<=right:
            mid = (left+right)//2
            if nums[mid] > nums[right]:
                left = mid +1
            elif nums[mid] < nums[right]:
                right = mid 

            else:
                 return nums[mid]
                

    
        
```

209. Minimum Size Subarray Sum

Given an array of positive integers nums and a positive integer target, return the minimal length of a subarray whose sum is greater than or equal to target. If there is no such subarray, return 0 instead.

 

Example 1:

Input: target = 7, nums = [2,3,1,2,4,3]
Output: 2
Explanation: The subarray [4,3] has the minimal length under the problem constraint.
Example 2:

Input: target = 4, nums = [1,4,4]
Output: 1
Example 3:

Input: target = 11, nums = [1,1,1,1,1,1,1,1]
Output: 0

```python
class Solution:
    def minSubArrayLen(self, target: int, nums: List[int]) -> int:
        left = 0
        current_sum = 0
        ans = float('inf')
        for right in range(len(nums)):
            current_sum +=nums[right]
            while current_sum >= target:
                ans = min(ans, right-left+1)

                current_sum -= nums[left]
                left+=1
        if ans == float('inf'):
            return 0
        return ans
        
```

## 268. Missing Number

Given an array nums containing n distinct numbers in the range [0, n], return the only number in the range that is missing from the array.

 

Example 1:

Input: nums = [3,0,1]

Output: 2

Explanation:

n = 3 since there are 3 numbers, so all numbers are in the range [0,3]. 2 is the missing number in the range since it does not appear in nums.

Example 2:

Input: nums = [0,1]

Output: 2

Explanation:

n = 2 since there are 2 numbers, so all numbers are in the range [0,2]. 2 is the missing number in the range since it does not appear in nums.

Example 3:

Input: nums = [9,6,4,2,3,5,7,0,1]

Output: 8

Explanation:

n = 9 since there are 9 numbers, so all numbers are in the range [0,9]. 8 is the missing number in the range since it does not appear in nums.

```python
class Solution:
    def missingNumber(self, nums: List[int]) -> int:
        sum = 0
        array_sum = 0
        for i in range(len(nums)):
            sum += (i+1)

            array_sum += nums[i]
        
        diff = sum - array_sum
        return diff


        

        
```
