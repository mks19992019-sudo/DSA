70. Climbing Stairs

You are climbing a staircase. It takes n steps to reach the top.

Each time you can either climb 1 or 2 steps. In how many distinct ways can you climb to the top?

Problem: Climbing Stairs

Initial Thought Process:
- I started by observing that to reach the nth stair, the last move can only come from:
  - stair (n - 1) by taking 1 step
  - stair (n - 2) by taking 2 steps
- This leads to the recurrence relation:

  f(n) = f(n - 1) + f(n - 2)

- Base cases:
  - f(1) = 1
  - f(2) = 2

First Attempt:
- I implemented a recursive solution directly using the recurrence relation.
- Although the logic was correct, it resulted in repeated computation of the same subproblems and exceeded the time limit for larger inputs.

Observation:
- While tracing the recursion tree, I noticed that values such as f(3), f(4), etc., were being calculated multiple times.
- This revealed the presence of overlapping subproblems, making the problem a good candidate for Dynamic Programming.

Optimization:
- I introduced memoization using a dictionary.
- Before computing a state, I check whether it has already been solved.
- If the result exists in the memo dictionary, it is returned immediately.
- Otherwise, the result is computed, stored, and reused whenever needed.

Key Mistake:
- Initially, I called self.climbStairs(n - 1) and self.climbStairs(n - 2) inside the memoized helper function.
- This bypassed the memoization logic.
- The correct approach is to call the helper function itself (dfs/rc) so that previously computed states can be reused.

Time Complexity: O(n)
Space Complexity: O(n)

```python
class Solution:
    def climbStairs(self, n: int) -> int:
        memo = {}
        def rc(n):
            if n == 1:
                return 1
            if n == 2:
                return 2
            if n in memo:
                return memo[n]
            
            memo[n] =  rc(n-1) + rc(n-2)

            return memo[n]

        return rc(n)


        
```
