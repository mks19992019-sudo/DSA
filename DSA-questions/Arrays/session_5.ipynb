70. Climbing Stairs

You are climbing a staircase. It takes n steps to reach the top.

Each time you can either climb 1 or 2 steps. In how many distinct ways can you climb to the top?

Problem: Climbing Stairs

Initial Thought Process:
- I started by observing that to reach the nth stair, the last move can only come from:
  - stair (n - 1) by taking 1 step
  - stair (n - 2) by taking 2 steps
- This leads to the recurrence relation:

  f(n) = f(n - 1) + f(n - 2)

- Base cases:
  - f(1) = 1
  - f(2) = 2

First Attempt:
- I implemented a recursive solution directly using the recurrence relation.
- Although the logic was correct, it resulted in repeated computation of the same subproblems and exceeded the time limit for larger inputs.

Observation:
- While tracing the recursion tree, I noticed that values such as f(3), f(4), etc., were being calculated multiple times.
- This revealed the presence of overlapping subproblems, making the problem a good candidate for Dynamic Programming.

Optimization:
- I introduced memoization using a dictionary.
- Before computing a state, I check whether it has already been solved.
- If the result exists in the memo dictionary, it is returned immediately.
- Otherwise, the result is computed, stored, and reused whenever needed.

Key Mistake:
- Initially, I called self.climbStairs(n - 1) and self.climbStairs(n - 2) inside the memoized helper function.
- This bypassed the memoization logic.
- The correct approach is to call the helper function itself (dfs/rc) so that previously computed states can be reused.

Time Complexity: O(n)
Space Complexity: O(n)

```python
class Solution:
    def climbStairs(self, n: int) -> int:
        memo = {}
        def rc(n):
            if n == 1:
                return 1
            if n == 2:
                return 2
            if n in memo:
                return memo[n]
            
            memo[n] =  rc(n-1) + rc(n-2)

            return memo[n]

        return rc(n)


        
```

171. Excel Sheet Column Number

Given a string columnTitle that represents the column title as appears in an Excel sheet, return its corresponding column number.

For example:

A -> 1
B -> 2
C -> 3
...
Z -> 26
AA -> 27
AB -> 28 
...
 

Example 1:

Input: columnTitle = "A"
Output: 1

solution: approach -> Convert the column title to a number by treating it as a base-26 number system.

```python
class Solution:
    def titleToNumber(self, columnTitle: str) -> int:
        ans = 0

        for ch in columnTitle:

            value = ord(ch) - ord("A") + 1
            ans = ans*26 + value

        return ans
```

202. Happy Number

Write an algorithm to determine if a number n is happy.

A happy number is a number defined by the following process:

Starting with any positive integer, replace the number by the sum of the squares of its digits.
Repeat the process until the number equals 1 (where it will stay), or it loops endlessly in a cycle which does not include 1.
Those numbers for which this process ends in 1 are happy.
Return true if n is a happy number, and false if not.

 

Example 1:

Input: n = 19
Output: true
Explanation:
12 + 92 = 82
82 + 22 = 68
62 + 82 = 100
12 + 02 + 02 = 1

```python
class Solution:
    def isHappy(self, n: int) -> bool:
        number = n
        sum = 0 
        memo = set()

        while number > 0:
            sum = 0
            while number>0:
                digit = number%10
                sum = digit*digit + sum
                number = number//10
            if sum ==1:
                return True
            
            if sum in memo:
                return False  
            else:
                memo.add(sum)
            number = sum
            

        
```

231. Power of Two

Given an integer n, return true if it is a power of two. Otherwise, return false.

An integer n is a power of two, if there exists an integer x such that n == 2x.

 

Example 1:

Input: n = 1
Output: true
Explanation: 20 = 1
Example 2:

Input: n = 16
Output: true
Explanation: 24 = 16
Example 3:

Input: n = 3
Output: false

```python

class Solution:
    def isPowerOfTwo(self, n: int) -> bool:
        if n<=0:
            return False
        while n>1:
            if n%2 !=0:
                return False
            n = n//2

        return True
            
```
