implement queue using stack

```python
class queueStack:
    def __init__(self):
        self.stk1 = []
        self.stk2 = []

    def push(self,x):
        while self.stk1:
            self.stk2.append(self.stk1.pop())

        self.stk1.append(x)

        while self.stk2:
            self.stk1.append(self.stk2.pop())


    def pop(self):
        if not self.stk1:
            return 'empty stack'
        
        return self.stk1.pop()
    
    def peek(self):
        if not self.stk1:
            return print('stack is empty')
        while self.stk1:
            self.stk2.append(self.stk1.pop())
        val = self.stk2[-1]

        while self.stk2:
            self.stk1.append(self.stk2.pop())
        return val
        


        


            
```

```python
q = queueStack()
```

```python
q.push(9)
```
