```python
def merge_sort(arr):
    if len(arr)==1:
        return arr
    n = len(arr)//2

    a1 = merge_sort(arr[:n])
    a2 = merge_sort(arr[n:])

    i = 0
    j = 0
    n = len(a1)
    m = len(a2)
    ans = []
    while i<n and j<m:
        if a1[i]<a2[j]:
            ans.append(a1[i])
            i+=1
        else:
            ans.append(a2[j])
            j+=1

    if i<n:
        ans.extend(a1[i:])

    else:
        ans.extend(a2[j:])
    return ans
```

```python
merge_sort([2,221,9,3,1002,8,33,5])
```
