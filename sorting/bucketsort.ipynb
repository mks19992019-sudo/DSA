```python
arr = [12, 45, 23, 67, 34, 89]

buckets = {
    "12-31": [],
    "32-51": [],
    "52-71": [],
    "72-91": []
}

for x in arr:
    if 12 <= x <= 31:
        buckets["12-31"].append(x)

    elif 32 <= x <= 51:
        buckets["32-51"].append(x)

    elif 52 <= x <= 71:
        buckets["52-71"].append(x)

    elif 72 <= x <= 91:
        buckets["72-91"].append(x)
```

```python
buckets

## function of bubble sort
def bubble_sort_dist(buckets):
    for key in buckets:
        print(key)
        # bubble sort
        for i in range(len(buckets[key])-1):
            for j in range(len(buckets[key])-1-i):
                if buckets[key][j] > buckets[key][j+1]:
                    temp = buckets[key][j]
                    buckets[key][j] = buckets[key][j+1]
                    buckets[key][j+1] = temp
    return buckets

print(buckets)
bubble_sort_dist(buckets)
```

```python
def bubble_sort_list(buckets):

    for arr in buckets:
        for i in range(len(arr)-1):
            for j in range(len(arr)-i-1):
                if arr[j] > arr[j+1]:
                    temp = arr[j]
                    arr[j] = arr[j+1]
                    arr[j+1] = temp
    return buckets
```
