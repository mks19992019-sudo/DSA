```python
arr = [2,1,9,3,8,5]
```

```python
for i in range(len(arr)-1):
    for j in range(i+1,len(arr)):
        if arr[i] > arr[j]:
            temp = arr[i]
            arr[i] = arr[j]
            arr[j] = temp
    print(arr)
```

```python
arr2 = [2,1,9,3,8,100]
```

```python
for i in range(len(arr2)-1):
    for j in range(len(arr2)-i-1):
        if arr2[j] > arr2[j+1]:
            temp = arr2[j]
            arr2[j] = arr2[j+1]
            arr2[j+1] = temp
    print(arr2)
```
