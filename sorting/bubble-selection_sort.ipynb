```python
arr = [2,4,1,5,7,4]
```

```python
n = len(arr)
for j in range(n):
    for i in range(n-1-j):
        if arr[i] > arr[i+1]:
            temp = arr[i]
            arr[i] = arr[i+1]
            arr[i+1] = temp

```

## selection sort

```python
for i in range(n-1):
    j_min = i+1

    for j in range(i+1,n):
        if arr[j_min] > arr[j]:
            j_min = j

    if arr[i] > arr[j_min]:
        temp = arr[i]
        arr[i] = arr[j_min]
        arr[j_min] = temp

        

arr

```
